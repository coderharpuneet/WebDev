# Matrix Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Heat Map

```python
flights=sns.load_dataset('flights')
flights
```

```python
tips=sns.load_dataset('tips')
```

```python
tipscorr=tips[['total_bill', 'tip', 'size']]
```

```python
tipscorr
```

```python
tips.corr()
```

```python
sns.heatmap(tips.corr(),cmap='coolwarm',annot=True)
```

## Cluster Map

```python
sns.clustermap(tips.corr(),cmap='coolwarm',annot=True)
```

## Pivot table with heatmap

```python
pvtflights=flights.pivot_table(index='month',columns='year',values='passengers')
pvtflights
```

```python
sns.heatmap(pvtflights,cmap='coolwarm',annot=True)
```

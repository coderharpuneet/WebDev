# Pandas Series

A series is a one dimensional labelled array capable of holding any data type. This axis labels are collectively called the index.

## Creating Series

There are multiple ways to create a pandas series.

```python
import numpy as np
import pandas as pd
```

```python
labels=['a','b','c']
my_list=[10,20,30]
arr=np.array([10,20,30])
d={1:10,2:20,3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list,index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```

# Data Frames

```python
# Creating a Data Frame

data={
     'Name' : ['John', 'Anna', 'Peter', 'Linda'],
     'Age' : [28,34,29,42],
     'City' : ['New York', 'Paris', 'Berlin', 'London'],
     'Salary' : [65000,70000,62000,85000]
}
df=pd.DataFrame(data)
print(df)
```

```python
data_list=[
     ['John',28,'New York',65000],
     ['Anna',34,'Paris',70000],
     ['Peter',29,'Berlin',62000],
     ['Linda',42,'London',85000]
]
df2=pd.DataFrame(data_list)
columns=["Name","Age","City","Salary"]
df2=pd.DataFrame(data_list,columns=columns)
print(df2)
```

```python
# Selection and indexing of columns

df2['Name']
```

```python
df2[["Name","City"]]
```

```python
# Creating a new column
df2["Designation"]=["Doctor","Eng.","Lawyer","Eng."]
```

```python
# Selecting subsets of rows and columns
df2.loc[[0,1]][["City","Salary"]]
```

```python
# Conditional Selection

# i only want to see whoe age is above 30
df2[df2["Age"]>30]
```

```python
# I only want people whose age is above 30 and their city must be paris
df2[(df2["Age"]>30) & (df2["City"]=="Paris")]
```

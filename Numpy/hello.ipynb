### Array Operations

```python
#Arithmetic Operations
a1=np.array([1,2,3,4,5])
a2=np.array([6,7,8,9,10])

summ=a1+a2
print(summ)
subb=a1-a2
print(subb)
mul=a1*a2
print(mul)
divv=a1/a2
print(divv)
print(a1//a2)
```

```python
# Broadcasting
l=[10,20,30,40,50]
a=np.array(l)
print(a+10)
```

```python
# Deep and shallow copy
a=np.arange(1,21)
a
```

```python
slice=a[:5]
print(slice*9)
```

```python
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])

# To take product of two matrices
product=np.dot(A,B)
print(product)
```

```python
#Transpose of a matrix
transpose=np.transpose(A)
print(transpose)
#OR
transpose=A.T
print(transpose)
```

```python
A=np.array([1,2,3,4,5])
B=np.array([6,7,8,9,10])
#Stacking of arrays
# For stacking we need to use vectors we cannot use matrices
stack=np.stack((A,B),axis=0)
print(stack) # Horizontal stacking
stack=np.stack((A,B),axis=1)
print(stack) # Vertical stacking
```

```python
# 2nd method
np.vstack((A,B))
```

```python
np.hstack((A,B))
```

```python
np.column_stack((A,B))
```
